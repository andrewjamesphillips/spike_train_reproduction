# Build a map from electrodes to preprocessing patterns

```python
# Imports
import numpy as np
import os
from tqdm import tqdm
from scipy.io import loadmat
import re
```

```python
preprocess_path = f"/Volumes/Stream/Analysis/RETINA/2024-10-02-7/data013"

for frame in os.listdir(preprocess_path):

    print(f"{frame}")
    pattern_path = os.path.join(preprocess_path, frame, "pattern_files")
    electrodes_to_preprocess_patterns = [[] for _ in range(519)]

    # Find the stimulating electrodes for each preprocessing pattern
    for file in tqdm(os.listdir(pattern_path)):

        # Extract the preprocessing pattern from the file using regular expression
        pp_pattern = int(re.findall('\d+', file)[0])

        # Load the mat file
        mat = loadmat(os.path.join(pattern_path, file))

        # Get the stimulating electrodes
        stim_elecs = mat['patternStruct']['stimElecs'][0,0][0,:]

        for elec in stim_elecs:
            electrodes_to_preprocess_patterns[elec-1].append(pp_pattern)

    # Convert the list of lists to a numpy array
    electrodes_to_preprocess_patterns = np.array(electrodes_to_preprocess_patterns, dtype=object)

    # Save the numpy array
    np.save(os.path.join(preprocess_path, frame, "electrodes_to_preprocess_patterns.npy"), electrodes_to_preprocess_patterns, allow_pickle=True)
```
