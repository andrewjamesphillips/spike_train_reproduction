```python
import numpy as np
import pickle
from tqdm import tqdm
import os
from scipy.io import loadmat, savemat
import matplotlib.pyplot as plt
import cv2
import rawmovie as rm
from PIL import Image
```

```python
# Create a RawMovieReader object
rm_obj = rm.RawMovieReader("/Volumes/Data/Stimuli/imagenet/INbrownian/INbrownian_05s_2500_0A_045.rawMovie")
stimulus_tensor,_ = rm_obj.get_frame_sequence(60*0,60*300)

natural_scenes = stimulus_tensor[0::60,:,:,0]

# Move the first axis to the last
natural_scenes = np.moveaxis(natural_scenes, 0, -1)

# Scale the images to -0.48 to 0.48
natural_scenes = natural_scenes / 255 * 0.96 - 0.48

# Convert the images to 8-bit integers
natural_scenes = ((natural_scenes+0.5)*256).astype(np.uint8)
```

```python
for i in range(150):
    print(i)
    plt.imshow(natural_scenes[:,:,i], cmap='gray')
    plt.show()
```
